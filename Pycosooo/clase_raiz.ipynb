## Propiedad de la Radicación: **Raíz de un Cociente**

---
### 📖 1. Explicación Teórica

En matemática, un **cociente** es una división o, lo que es lo mismo, una **fracción**.

$$\sqrt[n]{\frac{a}{b}} = \frac{\sqrt[n]{a}}{\sqrt[n]{b}} \quad (con \ b \neq 0)$$

- 🔵 **Arriba (numerador):** la raíz va arriba.
- 🟠 **Abajo (denominador):** la raíz va abajo.

---
### 💡 2. Tres Ejemplos Paso a Paso (en $\mathbb{Q}$ con resultados lindos)

#### **Ejemplo 1: Raíz cuadrada de una fracción**
$$\sqrt{\frac{4}{9}} = \frac{\sqrt{4}}{\sqrt{9}} = \frac{2}{3}$$
- Repartimos la raíz: $\sqrt{4}$ arriba y $\sqrt{9}$ abajo.
- $\sqrt{4} = 2$ (porque $2 \times 2 = 4$).
- $\sqrt{9} = 3$ (porque $3 \times 3 = 9$).
- **Resultado:** **2/3** (dos tercios).

#### **Ejemplo 2: Raíz cúbica de una fracción**
$$\sqrt[3]{\frac{8}{27}} = \frac{\sqrt[3]{8}}{\sqrt[3]{27}} = \frac{2}{3}$$
- Repartimos la raíz cúbica: $\sqrt[3]{8}$ arriba y $\sqrt[3]{27}$ abajo.
- La raíz cúbica de 8 es 2 (porque $2 \times 2 \times 2 = 8$).
- La raíz cúbica de 27 es 3 (porque $3 \times 3 \times 3 = 27$).
- **Resultado:** **2/3** (dos tercios).

#### **Ejemplo 3: Raíz cuadrada con números más grandes**
$$\sqrt{\frac{25}{36}} = \frac{\sqrt{25}}{\sqrt{36}} = \frac{5}{6}$$
- Distribuimos la raíz: $\sqrt{25}$ arriba y $\sqrt{36}$ abajo.
- $\sqrt{25} = 5$ y $\sqrt{36} = 6$.
- **Resultado:** **5/6** (cinco sextos).

```python
from fractions import Fraction

EJERCICIOS = [
    {"index": 2, "num": 1,   "den": 4,   "ans": Fraction(1, 2)},
    {"index": 2, "num": 9,   "den": 16,  "ans": Fraction(3, 4)},
    {"index": 2, "num": 25,  "den": 49,  "ans": Fraction(5, 7)},
    {"index": 2, "num": 1,   "den": 9,   "ans": Fraction(1, 3)},
    {"index": 2, "num": 16,  "den": 25,  "ans": Fraction(4, 5)},
    {"index": 2, "num": 36,  "den": 81,  "ans": Fraction(2, 3)},
    {"index": 2, "num": 49,  "den": 64,  "ans": Fraction(7, 8)},
    {"index": 2, "num": 4,   "den": 25,  "ans": Fraction(2, 5)},
    {"index": 2, "num": 9,   "den": 100, "ans": Fraction(3, 10)},
    {"index": 2, "num": 81,  "den": 100, "ans": Fraction(9, 10)},
    {"index": 3, "num": 1,   "den": 8,   "ans": Fraction(1, 2)},
    {"index": 3, "num": 8,   "den": 27,  "ans": Fraction(2, 3)},
    {"index": 3, "num": 27,  "den": 64,  "ans": Fraction(3, 4)},
    {"index": 3, "num": 1,   "den": 125, "ans": Fraction(1, 5)},
    {"index": 3, "num": 64,  "den": 125, "ans": Fraction(4, 5)},
    {"index": 2, "num": 100, "den": 9,   "ans": Fraction(10, 3)},
    {"index": 2, "num": 121, "den": 144, "ans": Fraction(11, 12)},
    {"index": 2, "num": 64,  "den": 4,   "ans": Fraction(4, 1)},
    {"index": 2, "num": 1,   "den": 36,  "ans": Fraction(1, 6)},
    {"index": 3, "num": 1,   "den": 27,  "ans": Fraction(1, 3)},
]

def format_rad(idx, n, d):
    return f"√({n}/{d})" if idx == 2 else f"3√({n}/{d})"

def parse_resp(txt):
    txt = txt.strip().replace(' ', '')
    if not txt:
        return None
    try:
        if '/' in txt:
            p = txt.split('/')
            return Fraction(int(p[0]), int(p[1])) if len(p) == 2 and int(p[1]) != 0 else None
        return Fraction(int(txt), 1)
    except ValueError:
        return None

print('=' * 50)
print('   RESOLVÉ LOS 20 EJERCICIOS INTERACTIVOS')
print('=' * 50)

total = len(EJERCICIOS)
for i, ej in enumerate(EJERCICIOS, 1):
    exp = format_rad(ej['index'], ej['num'], ej['den'])
    while True:
        r = input(f"Ejercicio {i}/{total} -> Calcula: {exp} = ").strip()
        p = parse_resp(r)
        if p is None:
            print('⚠️ Formato no válido. Escribe como fracción (ej: 2/3) o entero (ej: 4).')
            continue
        if p == ej['ans']:
            print('🎉 ¡Está bien! ¡Excelente trabajo!\n')
            break
        else:
            print('❌ Está mal. ¡Intenta otra vez! Revisa tus cuentas.\n')

print('🏆 ¡FELICITACIONES! Completaste los 20 ejercicios correctamente.')
```
