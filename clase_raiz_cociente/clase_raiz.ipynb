### 📖 1. Explicación Teórica

La propiedad establece que la raíz de una división o fracción se puede calcular distribuyendo el radical al numerador y al denominador:

$$\sqrt[n]{\frac{a}{b}} = \frac{\sqrt[n]{a}}{\sqrt[n]{b}}$$

**Conjuntos numéricos a los que pertenecen:**
- **$a \in \mathbb{Z}$** $\rightarrow$ **$a$** pertenece al conjunto de los números **enteros**.
- **$b \in \mathbb{Z} - \{0\}$** $\rightarrow$ **$b$** pertenece al conjunto de los números **enteros excluyendo el 0**.
- **$n \in \mathbb{N}$ ($n \ge 2$)** $\rightarrow$ **$n$** pertenece al conjunto de los números **naturales**.

*Nota: Si el índice $n$ es par, el radicando debe ser no negativo ($\frac{a}{b} \ge 0$). Si el índice $n$ es impar, el radicando puede ser positivo o negativo.*

---
### 💡 2. Tres Ejemplos Paso a Paso (en $\mathbb{Q}$ con resultados lindos)

#### **Ejemplo 1: Raíz cuadrada de una fracción**
$$\sqrt{\frac{4}{9}} = \frac{\sqrt{4}}{\sqrt{9}} = \frac{2}{3}$$
- Repartimos la raíz: $\sqrt{4}$ en el numerador y $\sqrt{9}$ en el denominador.
- $\sqrt{4} = 2$ (porque $2 \times 2 = 4$).
- $\sqrt{9} = 3$ (porque $3 \times 3 = 9$).
- **Resultado:** **2/3** (dos tercios).

#### **Ejemplo 2: Raíz cúbica de una fracción positiva**
$$\sqrt[3]{\frac{8}{27}} = \frac{\sqrt[3]{8}}{\sqrt[3]{27}} = \frac{2}{3}$$
- Repartimos la raíz cúbica: $\sqrt[3]{8}$ en el numerador y $\sqrt[3]{27}$ en el denominador.
- La raíz cúbica de 8 es 2 (porque $2 \times 2 \times 2 = 8$).
- La raíz cúbica de 27 es 3 (porque $3 \times 3 \times 3 = 27$).
- **Resultado:** **2/3** (dos tercios).

#### **Ejemplo 3: Raíz cúbica de una fracción negativa**
$$\sqrt[3]{-\frac{8}{27}} = \frac{\sqrt[3]{-8}}{\sqrt[3]{27}} = \frac{-2}{3} = -\frac{2}{3}$$
- Repartimos la raíz cúbica: $\sqrt[3]{-8}$ en el numerador y $\sqrt[3]{27}$ en el denominador.
- $\sqrt[3]{-8} = -2$ (porque $(-2) \times (-2) \times (-2) = -8$).
- $\sqrt[3]{27} = 3$ (porque $3 \times 3 \times 3 = 27$).
- **Resultado:** **-2/3** (menos dos tercios).
